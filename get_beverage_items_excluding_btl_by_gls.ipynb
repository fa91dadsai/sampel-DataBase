```python
import pandas as pd
import os
import re
```

```python
# read the excel file
path = r"C:\Users\fadia\OneDrive\Desktop\Ody\Data\odyssey_subcategories.xlsx"
# there is no  'sheet1' in the file, changed to correct sheet name
df = pd.read_excel(path)
df.head()
```

```python
# i want to filter the dataframe to get only where 'Categories' is 'Beverage'
df_beverage = df[df['Categories'] == 'Beverage']
# capitalize the columns names
df_beverage.columns = [col.capitalize() for col in df_beverage.columns]
# capitalize the items in 'Item' column
df_beverage['Item'] = df_beverage['Item'].str.title()
df_beverage.head()
```

```python
# btl items 
btl_path = r"C:\Users\fadia\OneDrive\Desktop\Ody\Data\beverage_btl_items.xlsx"
df_btl = pd.read_excel(btl_path)
df_btl.columns = [col.capitalize() for col in df_btl.columns]
# capitalize the items in 'Item' column
df_btl['Item'] = df_btl['Item'].str.title()
df_btl.head()
```

```python
# i want to exclude items that is in df_btl from df_beverage
beverage_items_gls = df_beverage[~df_beverage['Item'].isin(df_btl['Item'])] 
beverage_items_gls = beverage_items_gls[['Item', 'Sub-categories']]
beverage_items_gls.reset_index(drop=True, inplace=True)
beverage_items_gls.sort_values(by='Item', inplace=True)
beverage_items_gls
```

```python
# save to excel file
output_path = r"C:\Users\fadia\OneDrive\Desktop\Ody\Data\beverage_items_excluding_btl_by_gls.xlsx"
if not os.path.exists(os.path.dirname(output_path)):
    os.makedirs(os.path.dirname(output_path))
beverage_items_gls.to_excel(output_path, index=False)
```
